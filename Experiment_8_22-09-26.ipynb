```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('AirPassengers.csv')

print(df.columns)
print(df.head())
```

```python
df['Month'] = pd.to_datetime(df['Month'])
df = df.set_index('Month')

df.head()
```

```python
df.shape
```

```python
df.sample(5, random_state=0)
```

```python
df.loc['1949']
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(x=df.index, y=df['#Passengers'])

plt.xlabel('Date')
plt.ylabel('Passengers')
plt.title('Airline Passengers Time Series')

plt.show()
```

```python
df['Cumulative_Passengers'] = df['#Passengers'].cumsum()

df.head()
```

```python
plt.figure(figsize=(12,5))

sns.lineplot(x=df.index, y=df['Cumulative_Passengers'])

plt.xlabel('Date')
plt.ylabel('Cumulative Passengers')
plt.title('Cumulative Sum of Airline Passengers')

plt.show()
```

```python
yearly_mean = df['#Passengers'].resample('YS').mean()

yearly_mean
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    df['#Passengers'],
    marker='.',
    linestyle='-',
    linewidth=0.5,
    label='Monthly'
)

plt.plot(
    yearly_mean,
    marker='*',
    markersize=8,
    linestyle='-',
    label='Yearly Mean'
)

plt.xlabel('Date')
plt.ylabel('Passengers')
plt.title('Monthly Passengers vs Yearly Mean')
plt.legend()

plt.show()
```
