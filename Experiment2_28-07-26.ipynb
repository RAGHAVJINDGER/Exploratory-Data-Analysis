```python
#Imputation techniques

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer
```

```python
# Create sample DataFrame with missing values

df = pd.DataFrame({
    "Area": [1200, np.nan, 1500, 1800, 1000],
    "Price": [5000000, 6000000, np.nan, 8000000, 4500000],
    "Type": ["Flat", "Villa", "Flat", np.nan, "Villa"]
})

df
```

```python
# 1. Mean Imputation (Numerical)

mean_imputer = SimpleImputer(strategy="mean")

df["Area_Mean"] = mean_imputer.fit_transform(
    df[["Area"]]
)

df
```

```python
# 2. Median Imputation
# Numerical - robust to outliers

median_imputer = SimpleImputer(strategy="median")

df["Price_Median"] = median_imputer.fit_transform(
    df[["Price"]]
)

df
```

```python
# 3. Mode Imputation
# Categorical - most frequent

mode_imputer = SimpleImputer(strategy="most_frequent")

df["Type_Mode"] = mode_imputer.fit_transform(
    df[["Type"]]
).ravel()

df
```

```python
# 4. MICE Imputation
# Multivariate Imputation by Chained Equations

num_cols = ["Area", "Price"]

mice_imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

df_mice = df[num_cols].copy()

df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)

df["Area_MICE"] = df_mice["Area"]

df["Price_MICE"] = df_mice["Price"]

print(df)
```
