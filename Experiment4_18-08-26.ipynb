```python
# Bar chart of wine quality

df["quality"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Number of Wines by Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Number of Wines")
plt.xticks(rotation=0)

plt.show()
```

```python
# Distribution plot of alcohol

plt.figure(figsize=(10, 6))

sns.histplot(
    df["alcohol"],
    kde=True
)

plt.title("Distribution of Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.show()
```

```python
# Distribution of wine quality

plt.figure(figsize=(10, 6))

sns.histplot(
    df["quality"],
    kde=True,
    discrete=True
)

plt.title("Distribution of Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Frequency")

plt.show()
```

```python
# Box plot of alcohol

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["alcohol"]
)

plt.title("Box Plot of Alcohol")
plt.xlabel("Alcohol")

plt.show()
```

```python
# Box plot of wine quality

plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["quality"]
)

plt.title("Box Plot of Wine Quality")
plt.xlabel("Quality")

plt.show()
```

```python
# Scatter plot: Alcohol vs Quality

plt.figure(figsize=(10, 6))

plt.scatter(
    df["alcohol"],
    df["quality"]
)

plt.title("Scatter Plot of Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")

plt.show()
```

```python
# Scatter plot: Volatile Acidity vs Quality

plt.figure(figsize=(10, 6))

plt.scatter(
    df["volatile acidity"],
    df["quality"]
)

plt.title("Scatter Plot of Volatile Acidity vs Quality")
plt.xlabel("Volatile Acidity")
plt.ylabel("Quality")

plt.show()
```

```python
# Box plot of alcohol for different quality levels

plt.figure(figsize=(10, 6))

sns.boxplot(
    x="quality",
    y="alcohol",
    data=df
)

plt.title("Alcohol Distribution for Different Wine Quality Levels")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol")

plt.show()
```

```python
# Box plot of volatile acidity for different quality levels

plt.figure(figsize=(10, 6))

sns.boxplot(
    x="quality",
    y="volatile acidity",
    data=df
)

plt.title("Volatile Acidity for Different Wine Quality Levels")
plt.xlabel("Wine Quality")
plt.ylabel("Volatile Acidity")

plt.show()
```

```python
# Pair plot

selected_columns = [
    "fixed acidity",
    "volatile acidity",
    "citric acid",
    "alcohol",
    "quality"
]

sns.pairplot(
    df[selected_columns]
)

plt.show()
```

```python
# Pearson correlation between alcohol and quality

corr = stats.pearsonr(
    df["alcohol"],
    df["quality"]
)

print("p-value:", corr[1])
print("Correlation coefficient:", corr[0])
```

```python
# Pearson correlation between volatile acidity and quality

corr = stats.pearsonr(
    df["volatile acidity"],
    df["quality"]
)

print("p-value:", corr[1])
print("Correlation coefficient:", corr[0])
```

```python
# Pearson correlation between alcohol and sulphates

corr = stats.pearsonr(
    df["alcohol"],
    df["sulphates"]
)

print("p-value:", corr[1])
print("Correlation coefficient:", corr[0])
```

```python
df.corr(method='pearson', numeric_only=True)
```

```python
# Pearson correlation matrix

correlation = df.corr(
    method="pearson",
    numeric_only=True
)

correlation
```

```python
# Correlation heatmap

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    xticklabels=correlation.columns,
    yticklabels=correlation.columns
)

plt.title("Correlation Heatmap of WineQT Dataset")

plt.show()
```

```python
# Find strongest correlations with wine quality

quality_corr = correlation["quality"].drop("quality")

strongest_positive = quality_corr.idxmax()
strongest_negative = quality_corr.idxmin()

print(
    "Strongest positive correlation with quality:",
    strongest_positive,
    "=",
    quality_corr[strongest_positive]
)

print(
    "Strongest negative correlation with quality:",
    strongest_negative,
    "=",
    quality_corr[strongest_negative]
)
```

```python
# Sorted correlation values with quality

quality_corr.sort_values(
    ascending=False
)
```

```python
# Complete correlation analysis

from scipy import stats

# Pearson correlation: alcohol vs quality
corr_alcohol_quality = stats.pearsonr(
    df["alcohol"],
    df["quality"]
)

print("Alcohol vs Quality")
print("Correlation:", corr_alcohol_quality[0])
print("P-value:", corr_alcohol_quality[1])

# Pearson correlation matrix
correlation = df.corr(
    method="pearson",
    numeric_only=True
)

print("\nCorrelation Matrix:")
display(correlation)

# Heatmap
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Pearson Correlation Heatmap - WineQT")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.show()# Final summary of WineQT dataset

print("========== WINEQT DATASET SUMMARY ==========")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:")
print(df.isna().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nColumns:")
print(df.columns.tolist())

print("\nDescriptive statistics:")
display(df.describe())
```
