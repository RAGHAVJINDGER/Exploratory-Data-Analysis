```python
import pandas as pd

columns = [
    'amt',
    'city_pop',
    'lat',
    'long',
    'merch_lat',
    'merch_long',
    'is_fraud'
]

train_df = pd.read_csv(
    '/content/fraud_data/CreditCard/fraudTrain.csv',
    usecols=columns
)

test_df = pd.read_csv(
    '/content/fraud_data/CreditCard/fraudTest.csv',
    usecols=columns
)

print("Training Dataset Shape:", train_df.shape)
print("Testing Dataset Shape:", test_df.shape)

train_df.head()
```

```python
# Exp. 5.1 - Discretization and Binning

amount = train_df['amt']

amount.head(10)
```

```python
bins = [0, 50, 100, 500, 1000, float('inf')]

category = pd.cut(
    amount,
    bins=bins,
    include_lowest=True
)

category
```

```python
pd.Series(category).value_counts().sort_index()
```

```python
bin_names = [
    'Very Low',
    'Low',
    'Medium',
    'High',
    'Very High'
]

amount_category = pd.cut(
    amount,
    bins=bins,
    labels=bin_names,
    include_lowest=True
)

amount_category
```

```python
pd.Series(amount_category).value_counts()
```

```python
pd.cut(
    amount,
    5,
    precision=2
)
```

```python
equal_bins = pd.cut(
    amount,
    5,
    precision=2
)

pd.Series(equal_bins).value_counts().sort_index()
```

```python
randomNumbers = train_df['amt'].sample(
    n=2000,
    random_state=42
)

category3 = pd.qcut(
    randomNumbers,
    4
)

category3
```

```python
pd.Series(category3).value_counts().sort_index()
```

```python
quartile_names = [
    'Q1 - Very Low',
    'Q2 - Low',
    'Q3 - High',
    'Q4 - Very High'
]

amount_quartile = pd.qcut(
    randomNumbers,
    4,
    labels=quartile_names
)

amount_quartile
```

```python
pd.Series(amount_quartile).value_counts()
```

```python
# Exp. 5.2 - Feature Scaling and Normalization

features = [
    'amt',
    'city_pop',
    'lat',
    'long',
    'merch_lat',
    'merch_long'
]

df_scaling = train_df[features].copy()

df_scaling.head()
```

```python
df_scaling.isnull().sum()
```

```python
df_scaling = df_scaling.dropna()

df_scaling.head()
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

min_max_scaler = MinMaxScaler()

df_min_max = pd.DataFrame(
    min_max_scaler.fit_transform(df_scaling),
    columns=df_scaling.columns
)

print("Min-Max Scaled Data:")
print(df_min_max.head())
```

```python
standard_scaler = StandardScaler()

df_standard = pd.DataFrame(
    standard_scaler.fit_transform(df_scaling),
    columns=df_scaling.columns
)

print("Standardized Data:")
print(df_standard.head())
```

```python
robust_scaler = RobustScaler()

df_robust = pd.DataFrame(
    robust_scaler.fit_transform(df_scaling),
    columns=df_scaling.columns
)

print("Robust Scaled Data:")
print(df_robust.head())
```

```python
outliers_1000 = train_df[
    np.abs(TotalTransaction) > 1000
]

print("Number of transactions above 1000:")
print(len(outliers_1000))
```

```python
outliers_5000 = train_df[
    np.abs(TotalTransaction) > 5000
]

print("Number of transactions above 5000:")
print(len(outliers_5000))
```

```python
Q1 = train_df["amt"].quantile(0.25)
Q3 = train_df["amt"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
iqr_outliers = train_df[
    (train_df["amt"] < lower_bound) |
    (train_df["amt"] > upper_bound)
]

iqr_outliers.head(10)
```

```python
print("Number of IQR Outliers:")
print(len(iqr_outliers))
```

```python
iqr_outliers["is_fraud"].value_counts()
```

```python
train_df["is_fraud"].value_counts()
```

```python
train_df.groupby("is_fraud")["amt"].describe()
```

```python

print("OUTLIER DETECTION")

print("\nTotal Transactions:")
print(len(train_df))

print("\nTransactions Above 1000:")
print(len(outliers_1000))

print("\nTransactions Above 5000:")
print(len(outliers_5000))

print("\nIQR Outliers:")
print(len(iqr_outliers))

print("\nFraud Distribution Among IQR Outliers:")
print(iqr_outliers["is_fraud"].value_counts())
```
